```python
import pandas as pd

data_path = "../src/data/raw/thyroid0387.data"
names_path = "../src/data/raw/thyroid0387.names"

with open(names_path, "r") as f:
    names_text = f.read()

print("=== .names file content ===")
print(names_text)

raw_df = pd.read_csv(data_path, header=None, sep=',|\s+', engine='python', dtype=str)
question_counts = (raw_df == "?").sum()

df = pd.read_csv(data_path, header=None, sep=',|\s+', engine='python', na_values="?", keep_default_na=False)
col_names = [f"feature_{i}" for i in range(df.shape[1] - 1)] + ["target"]
df.columns = col_names

print("\n=== Dataset shape ===")
print(df.shape)

print("\n=== First few rows ===")
print(df.head())

print("\n=== Data types ===")
print(df.dtypes)

print("\n=== Missing values count before replacing '?' ===")
print(question_counts)

print("\n=== Missing values count after replacing '?' with NaN ===")
print(df.isna().sum())

print("\n=== Target value counts ===")
print(df["target"].value_counts(dropna=False))

print("\n=== Number of unique values per column ===")
print(df.nunique(dropna=False))
```
